### Verify Semantic vs TF-IDF Feature Separation


```python
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df_pairs = pd.read_csv(PROJECT_ROOT / 'data' / 'features' / 'paper_patent_pairs.csv')

print(f'Total Candidate Pairs: {len(df_pairs)}')
sem = df_pairs['semantic_similarity']
tfidf = df_pairs['tfidf_similarity']

exact_matches = (sem == tfidf).sum()
corr = sem.corr(tfidf)
print(f'Exact Matches (sem == tfidf): {exact_matches}')
print(f'Pearson Correlation: {corr:.4f} (Confirms decoupled, complementary features)')
df_pairs[['semantic_similarity', 'tfidf_similarity', 'jaccard_similarity', 'label']].describe()
```

### Visualize Feature Space Separation


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))
plt.scatter(df_pairs['tfidf_similarity'], df_pairs['semantic_similarity'], c=df_pairs['label'], cmap='coolwarm', alpha=0.6)
plt.title(f'TF-IDF vs Semantic Similarity (r = {corr:.2f})')
plt.xlabel('TF-IDF Similarity')
plt.ylabel('Semantic Similarity (MiniLM-L6-v2)')
plt.colorbar(label='Ground Truth Label')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
