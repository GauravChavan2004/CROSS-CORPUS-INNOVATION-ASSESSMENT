### Load Processed Clean Data


```python
import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
data_proc = PROJECT_ROOT / 'data' / 'processed'

df_papers = pd.read_csv(data_proc / 'papers_clean.csv')
df_patents = pd.read_csv(data_proc / 'patents_clean.csv')

print(f'Clean Papers: {len(df_papers)} records')
print(f'Clean Patents: {len(df_patents)} records')
print(f'Candidate Cartesian Space: {len(df_papers) * len(df_patents)} pairs')
```
