Visualizes feature importances, ROC curves, and confusion matrices for the production classifier.


### Load and Render Production Figures


```python
from PIL import Image
import matplotlib.pyplot as plt
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
fig_dir = PROJECT_ROOT / 'reports' / 'figures'

for fig_name in ['feature_importance.png', 'confusion_matrix.png', 'tfidf_vs_semantic_distribution.png']:
    img_p = fig_dir / fig_name
    if img_p.exists():
        print(f'Displaying: {fig_name}')
        img = Image.open(img_p)
        plt.figure(figsize=(9, 4.5))
        plt.imshow(img)
        plt.axis('off')
        plt.show()
```
