### Plot Temporal Distribution of Corpus Documents


```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df_papers = pd.read_csv(PROJECT_ROOT / 'data' / 'processed' / 'papers_clean.csv')
df_patents = pd.read_csv(PROJECT_ROOT / 'data' / 'processed' / 'patents_clean.csv')

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(df_papers['publication_year'], bins=10, color='#3b82f6', edgecolor='black')
axes[0].set_title('Paper Publication Years')
axes[1].hist(df_patents['publication_year'], bins=10, color='#10b981', edgecolor='black')
axes[1].set_title('Patent Publication Years')
plt.tight_layout()
plt.show()
```
